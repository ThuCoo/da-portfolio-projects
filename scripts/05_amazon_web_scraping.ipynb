```python
# --- Fetching Website ---
# For demonstration purpose only
# Saving page as Website, HTML Only

# Read Local HTML File
filePath = r'C:\Users\THU THAO\Downloads\self studi\data analyst\data\Amazon\Amazon.html'
with open(filePath, 'r', encoding='utf-8') as file:
    soup = BeautifulSoup(file.read(), 'html.parser')
    
title = soup.find('span', id='productTitle').get_text(strip=True)
price = float(soup.find('span', class_='a-offscreen').get_text(strip=True)[1:])
# priceNum = float(re.search(r'\d+(.*)', price).group())
today = datetime.date.today()
```

```python
# Export to csv
header = ['Title', 'Price', 'CollectDate']
data = [title, price, today]

csvPath = r'C:\Users\THU THAO\Downloads\self studi\data analyst\data\Amazon\amazon.csv'
with open(csvPath, 'w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(header)
    writer.writerow(data)
```

```python
# Append to csv
csvPath = r'C:\Users\THU THAO\Downloads\self studi\data analyst\data\Amazon\amazon.csv'
with open(csvPath, 'a+', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(data)
```
